#### Auto Mpg

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso
import seaborn as sns
from sklearn import preprocessing
```

```python
df = pd.read_csv('auto-mpg.csv')
df.head()
```

```python
df.head()
df.shape
df.columns
df.dtypes
df.duplicated().sum()
df.nunique()
df.cylinders.nunique()
df.info()
df.describe()
df.isnull().sum()
df.corr(numeric_only = True)
df['cylinders'].value_counts()
df['origin'].value_counts().to_frame()
(df['horsepower'].str.isnumeric() == False).sum()
(df['horsepower'] == '?').sum()
df[df['horsepower'] == '?']

df[df['origin'] == 1].shape
df[df['origin'] == 2].shape
df[df['origin'] == 3].shape
```

#### Data Cleaning

```python
df = df.drop('car name', axis = 1)
df.head()
```

```python
df = df.replace('?', np.nan)
```

```python
df = df.apply(lambda x: x.fillna(x.mode()[0]), axis = 0)
df
```

```python
df.info()
```

```python
df['horsepower'] = df['horsepower'].astype(int)
df.info()
```

#### Create Model

```python
X = df.drop('mpg',  axis = 1)
y = df[['mpg']]
X.shape, y.shape
```

```python
# X_scaled = preprocessing.scale(X)
# X_scaled = pd.DataFrame(X_scaled, columns = X.columns)

# y_scaled = preprocessing.scale(y)
# y_scaled = pd.DataFrame(y_scaled, columns = y.columns)

# X_scaled.shape, y_scaled.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = .2, random_state = 42)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
data = {'y_Actual' : y_test, 'y_pred' : y_pred}

data

# df = pd.DataFrame(data, columns = ['y_Actual', 'y_pred'])
# df
```

```python
for i, col_name in enumerate(X_train.columns):
    print(f'Regression model coefficient for {col_name} =>\t\t\t{model.coef_[0][i]}')
    
print(f'intercept => {model.intercept_[0]}')
```

```python
ridge = Ridge(alpha = .2)
ridge.fit(X_train, y_train)
```

```python
for i, col_name in enumerate(X_train.columns):
    print(f'Regression model coefficient for {col_name} =>\t\t\t{ridge.coef_[i]}')
ridge.coef_ 
```

```python
print(ridge.score(X_train, y_train))
print(ridge.score(X_test, y_test))
```

```python
lasso = Lasso(alpha = .2)
lasso.fit(X_train, y_train)
```

```python
for i, col_name in enumerate(X_train.columns):
    print(f'Regression model coefficient for {col_name} =>\t\t\t{lasso.coef_[i]}')
lasso.coef_ 
```

```python
print(lasso.score(X_train, y_train))
print(lasso.score(X_test, y_test))
```
